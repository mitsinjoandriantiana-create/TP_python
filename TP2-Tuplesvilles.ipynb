# TP 2 — Tuples et villes de Madagascar

## Étape 1 — Coordonnées sous forme de tuples

```python
villes = {
    "Antananarivo": (-18.8792, 47.5079),
    "Mahajanga": (-15.7167, 46.3167),
    "Toamasina": (-18.1492, 49.4023),
    "Fianarantsoa": (-21.4536, 47.0857),
    "Toliara": (-23.3500, 43.6667),
    "Antsiranana": (-12.2787, 49.2917),
}
for nom, coord in villes.items():
    print(nom, coord)
```

## Étape 2 — Ville la plus au nord

```python
nord = max(villes, key=lambda nom: villes[nom][0])
print("Ville la plus au nord :", nord)
```

## Étape 3 — Fonction qui vérifie des limites

```python
def dans_limites(coord, lat_min, lat_max, lon_min, lon_max):
    lat, lon = coord
    return lat_min <= lat <= lat_max and lon_min <= lon <= lon_max

print(dans_limites(villes["Antananarivo"], -20, -17, 46, 49))
```

## Étape 4 — Distance plane avec math.dist

```python
from math import dist
print("Distance en degrés (pas en km) :",
      round(dist(villes["Antananarivo"], villes["Toamasina"]), 3))
```

## Étape 5 — Distances entre toutes les paires

```python
noms = list(villes)
for i in range(len(noms)):
    for j in range(i + 1, len(noms)):
        print(noms[i], "-", noms[j], ":", round(dist(villes[noms[i]], villes[noms[j]]), 3))
```

## Étape 6 — Immutabilité du tuple

```python
coord = villes["Antananarivo"]
try:
    coord[0] = -19
except TypeError as erreur:
    print("Un tuple ne peut pas être modifié :", erreur)
```

## Bonus — Distance Haversine en kilomètres

```python
from math import radians, sin, cos, asin, sqrt
def haversine(a, b):
    lat1, lon1 = map(radians, a)
    lat2, lon2 = map(radians, b)
    dlat, dlon = lat2-lat1, lon2-lon1
    h = sin(dlat/2)**2 + cos(lat1)*cos(lat2)*sin(dlon/2)**2
    return 6371 * 2 * asin(sqrt(h))

print(round(haversine(villes["Antananarivo"], villes["Toamasina"]), 1), "km")
```
