# TP 1 — Listes et notes

## Étape 1 — Données

```python
notes = [10, 14, 7, 18, 11, 9, 16, 13, 5, 15]
print("Notes :", notes)
print("Effectif :", len(notes))
```

## Étape 2 — Statistiques

```python
print("Moyenne :", round(sum(notes) / len(notes), 2))
print("Minimum :", min(notes))
print("Maximum :", max(notes))
print("Nombre de notes >= 10 :", sum(n >= 10 for n in notes))
```

## Étape 3 — Admis et classement

```python
admis = [n for n in notes if n >= 10]
classement = sorted(notes, reverse=True)
print("Admis :", admis)
print("Classement décroissant :", classement)
print("Top 3 :", classement[:3])
```

## Étape 4 — Délibération : +2 si la note est inférieure à 10, maximum 10

```python
deliberees = [min(n + 2, 10) if n < 10 else n for n in notes]
print("Notes après délibération :", deliberees)
```

## Étape 5 — Histogramme des tranches

```python
tranches = {"0-4": (0, 4), "5-9": (5, 9), "10-14": (10, 14), "15-20": (15, 20)}
for nom, (bas, haut) in tranches.items():
    effectif = sum(bas <= n <= haut for n in notes)
    print(f"{nom:6} {'█' * effectif} ({effectif})")
```

## Étape 6 — Rang de chaque élève et fusion sans doublons

```python
for rang, (numero, note) in enumerate(sorted(enumerate(notes, 1), key=lambda x: x[1], reverse=True), 1):
    print(f"{rang}. Élève {numero} : {note}/20")
autres_notes = [12, 8, 17, 20]
print("Fusion unique :", sorted(set(notes) | set(autres_notes)))
```
