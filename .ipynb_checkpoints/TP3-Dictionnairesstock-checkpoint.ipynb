# TP 3 — Dictionnaires et stock d'épicerie

## Étape 1 — Initialiser le stock

```python
stock = {
    "riz": {"quantite": 20, "prix": 3000},
    "huile": {"quantite": 12, "prix": 7000},
    "sucre": {"quantite": 15, "prix": 4000},
    "savon": {"quantite": 0, "prix": 2500},
    "farine": {"quantite": 10, "prix": 3500},
}
for nom, article in stock.items():
    print(nom, article)
```

## Étape 2 — Valeur totale et produits épuisés

```python
total = sum(a["quantite"] * a["prix"] for a in stock.values())
epuises = [nom for nom, a in stock.items() if a["quantite"] == 0]
print("Valeur du stock :", total, "Ar")
print("Épuisés :", epuises)
```

## Étape 3 — Fonction de vente

```python
def vendre(nom, quantite):
    article = stock.get(nom)
    if article is None:
        print("Produit inconnu"); return False
    if quantite <= 0:
        print("Quantité invalide"); return False
    if article["quantite"] < quantite:
        print("Stock insuffisant"); return False
    article["quantite"] -= quantite
    print("Vente réussie")
    return True

vendre("riz", 3)
vendre("savon", 1)
```

## Étape 4 — Ajouter un produit sans doublon

```python
def ajouter(nom, quantite, prix):
    if nom in stock:
        print("Produit déjà présent"); return False
    if quantite < 0 or prix < 0:
        print("Valeurs invalides"); return False
    stock[nom] = {"quantite": quantite, "prix": prix}
    return True

print(ajouter("café", 8, 6000))
```

## Étape 5 — Hausse de prix et comptage des ventes

```python
for article in stock.values():
    article["prix"] = round(article["prix"] * 1.10)
from collections import Counter
ventes = Counter()
ventes.update(["riz", "riz", "huile", "riz"])
print("Ventes :", dict(ventes))
print("Prix après hausse :", {n: a["prix"] for n, a in stock.items()})
```

## Étape 6 — Sauvegarde JSON

```python
import json
from pathlib import Path
fichier = Path("stock_ami.json")
fichier.write_text(json.dumps(stock, ensure_ascii=False, indent=2), encoding="utf-8")
copie = json.loads(fichier.read_text(encoding="utf-8"))
print("Fichier :", fichier.resolve())
print("Rechargement identique :", copie == stock)
```
